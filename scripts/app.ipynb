Librerías

```python
import csv
import os
from pathlib import Path
from pypdf import PdfReader, PdfWriter
```

Lista de alumnos

```python
alumnos = []

with open("alumnos_consolidad.csv", newline="", encoding="utf-8") as archivo:
    reader = csv.reader(archivo)

    # Saltar encabezado
    next(reader)

    for fila in reader:
        alumno = {
            "codigo": fila[0],
            "alumno": fila[1],
            "nombre": fila[2],
            "ec1": {
                "inicio": int(fila[3]),
                "fin": int(fila[4]),
            },
            "ec2": {
                "inicio": int(fila[5]),
                "fin": int(fila[6]),
            },
            "ec3": {
                "inicio": int(fila[7]),
                "fin": int(fila[8]),
            },
            "ee1": {
                "inicio": int(fila[9]),
                "fin": int(fila[10]),
            },
            "pro": {
                "inicio": int(fila[11]),
                "fin": int(fila[12]),
            },
            "ee2": {
                "inicio": int(fila[13]),
                "fin": int(fila[14]),
            },
        }

        alumnos.append(alumno)

print(alumnos)
```

Crear Carpetas

```python
def crear_carpetas(): 
    # Crear la carpeta principal
    tmp = Path("EC")
    tmp.mkdir(exist_ok=True)

    # Crear una carpeta por alumno usando la llave "nombre"
    for alumno in alumnos:
        carpeta = tmp / alumno["nombre"]
        carpeta.mkdir(exist_ok=True)

    print("Carpetas creadas.")

    # +++++++++++++++++++++++++++++++++++++++++++++++++++++++++++

    tmp = Path("EE1")
    tmp.mkdir(exist_ok=True)

    print("Carpetas creadas.")

    tmp = Path("PRO")
    tmp.mkdir(exist_ok=True)

    print("Carpetas creadas.")

    tmp = Path("EE2")
    tmp.mkdir(exist_ok=True)

    print("Carpetas creadas.")
```

```python
crear_carpetas()

pdfs = [
    ("EC1", PdfReader("EC1.pdf")),
    ("EC2", PdfReader("EC2.pdf")),
    ("EC3", PdfReader("EC3.pdf")),
]

tmp = Path("EC")

for alumno in alumnos:
    carpeta = tmp / alumno["nombre"]

    for nombre_ec, pdf in pdfs:
        writer = PdfWriter()

        rango = alumno[nombre_ec.lower()]  # ec1, ec2, ec3

        for i in range(rango["inicio"] - 1, rango["fin"]):
            writer.add_page(pdf.pages[i])

        salida = carpeta / f"{alumno['nombre']} - {nombre_ec}.pdf"

        with open(salida, "wb") as f:
            writer.write(f)

# +++++++++++++++++++++++++++++++++++++++++++++++

tmp = Path("EE1")

for alumno in alumnos:
    carpeta = "EE1"

    pdf = PdfReader("EE1.pdf")

    writer = PdfWriter()

    rango = alumno["ee1"]  # ec1, ec2, ec3

    for i in range(rango["inicio"] - 1, rango["fin"]):
        writer.add_page(pdf.pages[i])

    salida = os.path.join(carpeta, f"{alumno['nombre']}.pdf")

    with open(salida, "wb") as f:
        writer.write(f)

# +++++++++++++++++++++++++++++++++++++++++++++++

tmp = Path("PRO")

for alumno in alumnos:
    carpeta = "PRO"

    pdf = PdfReader("PRO.pdf")

    writer = PdfWriter()

    rango = alumno["pro"]  # ec1, ec2, ec3

    for i in range(rango["inicio"] - 1, rango["fin"]):
        writer.add_page(pdf.pages[i])

    salida = os.path.join(carpeta, f"{alumno['nombre']}.pdf")

    with open(salida, "wb") as f:
        writer.write(f)

# +++++++++++++++++++++++++++++++++++++++++++++++}

tmp = Path("PRO")

for alumno in alumnos:
    carpeta = "PRO"

    pdf = PdfReader("PRO.pdf")

    writer = PdfWriter()

    rango = alumno["pro"]  # ec1, ec2, ec3

    for i in range(rango["inicio"] - 1, rango["fin"]):
        writer.add_page(pdf.pages[i])

    salida = os.path.join(carpeta, f"{alumno['nombre']}.pdf")

    with open(salida, "wb") as f:
        writer.write(f)

# +++++++++++++++++++++++++++++++++++++++++++++++

tmp = Path("EE2")

for alumno in alumnos:
    carpeta = "EE2"

    pdf = PdfReader("EE2.pdf")

    writer = PdfWriter()

    rango = alumno["pro"]

    for i in range(rango["inicio"] - 1, rango["fin"]):
        writer.add_page(pdf.pages[i])

    salida = os.path.join(carpeta, f"{alumno['nombre']}.pdf")

    with open(salida, "wb") as f:
        writer.write(f)

    
```
